```python
# [CELL 1] STUDENT CONFIG

LAST_NAME = "Mahinay"
STUDENT_ID = "TUPM-26-4014"
SEED_NUM = int(STUDENT_ID[-1])
FAVORITE_ARTIST = "NIKI,KESHI,LANY,AVRILLAVIGNE,PARAMORE,CLAIRO,LAUFEY,MALCOLMTODD,SUYEN,CHASEATLANTIC,THEWEEKND"

print("Student:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
```

```python
# [CELL 2] FAULT CODE GENERATION

def identity_number():
    return sum(ord(char) for char in LAST_NAME + FAVORITE_ARTIST) + SEED_NUM
def generate_fault_code():
    number = identity_number()
    digits = ''.join(str((number // (index + 1)) % 10) for index in range(5))
    return f'F-{SEED_NUM}{digits}'
fault_code = generate_fault_code()
print('Generated Fault Code:', fault_code)
```

```python
# [CELL 3] RECURSIVE FAULT TRACE

def trace_fault(fault_code, depth, trace=None):
    if trace is None:
        trace = []
    trace.append(f'Level {depth}: {fault_code}')
    if depth == 0:
        trace.append('Root cause isolated: identity-specific diagnostic fault.')
        return trace, len(trace) - 1
    return trace_fault(fault_code, depth - 1, trace)
fault_trace, recursive_calls = trace_fault(fault_code, SEED_NUM + 2)
```

```python
# [CELL 4] DIAGNOSTIC REPORT

print('=== EXERCISE 2: RECURSIVE FAULT TRACE ===')
print(f'Student: {LAST_NAME} | Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}')
print(f'Fault code: {fault_code}')
print(f'Recursive calls: {recursive_calls}')
print('Complete diagnostic sequence:')
print('\n'.join(f'  {step}' for step in fault_trace))
print(f'Final result: {fault_trace[-1]}')
     
```
